### range()

```python
N = 10
for i in range(N): #i from 0 to N-1
    print(i)
```

`range(N)` generates `0, 1, ..., N-1` - N numbers total, but never reaching N itself. This is the most common way to loop a fixed number of times in Python, playing the same role as Java's `for (int i = 0; i < N; i++)`.

### range() With Step, and .format()

```python
for i in range(7, 2, -2):
    print(i)
    print("I said {}!".format(i))
```

`range(7, 2, -2)` adds a start, stop, and step: begin at 7, stop *before* reaching 2, and step by −2 each time (counting down) - giving `7, 5, 3`. `.format()` is an older way to build strings with placeholders (`{}`), doing the same job as the f-strings (`f"..."`) used elsewhere in these notebooks - both are fine, f-strings are just newer and slightly more concise.

### Nested Loops and O(n²)

```python
def unique(array):
    for i in range(len(array)):
        for j in range(len(array)):
            if i != j and array[i] == array[j]:
                return False
    return True
```

```python
L = [2, 4, 5, 6, 7, 4]
unique(L)
```

`unique` checks every pair of positions `(i, j)` (`i != j` just skips comparing an element to itself) and bails out the moment it finds a match. That's a nested loop - roughly `len(array)` iterations of the outer loop, each running the full inner loop again - which is the classic shape of an **O(n²)** algorithm: double the input size, and the work roughly quadruples.

```python
from time import time
import matplotlib.pyplot as plt
```

What is the time complexity of the function below?

### Timing Helper

```python
def time_unique(array, unique_func):
    start_time = time()
    unique_func(array)
    end_time = time()
    return end_time - start_time
```

`time_unique` is a small timing helper: it records the clock before and after calling whatever function is passed in (`unique_func`), and returns the elapsed time. This lets you *measure* an algorithm's speed empirically, rather than only reasoning about its Big-O on paper.

### Benchmarking unique()

```python
N = 2000
lengths = list(range(N))
times = [0 for _ in range(N)]

for arr_len in range(N):
    array = [x for x in range(arr_len)] #we cheat and make sure it is unique = worst case
    times[arr_len] = time_unique(array, unique)
        
```

This times `unique` on arrays of steadily increasing size (`arr_len` from 0 up to 2000), building already-unique arrays on purpose so the function is forced to scan the *entire* array every time (the worst case) rather than exiting early on a duplicate.

### Reading the Plot

```python
plt.figure()
plt.plot(lengths, times, '.b')
plt.show()
```

The resulting plot should curve noticeably upward rather than rise in a straight line - that upward curve is the visual signature of roughly quadratic (O(n²)) growth: as input size increases, time increases much faster than proportionally.

### Optimizing unique()

```python
def better_unique(array):
    for i in range(len(array)):
        for j in range(i+1, len(array)):
            if array[i] == array[j]:
                return False
    return True
```

`better_unique` only compares each pair *once*, starting `j` right after `i` instead of from the beginning every time - roughly half the total comparisons of the first version. Worth noting: this is still O(n²) in Big-O terms (a constant-factor speedup, not a different growth category) - it'll be noticeably faster in practice, but the curve shape as input size grows should look similar.

```python
times_better = [0 for _ in range(N)]

for arr_len in range(N):
    array = [x for x in range(arr_len)] #we cheat and make sure it is unique = worst case
    times_better[arr_len] = time_unique(array, better_unique)
      
```

### Comparing the Two Curves

```python
plt.figure()
plt.plot(lengths, times, '.b', label = "unique")
plt.plot(lengths, times_better, '.r', label = "better_unique")
plt.legend()
plt.xlabel("Input size, n")
plt.ylabel("Time, t")
plt.show()
```

Comparing the two curves side by side is the point: both should still bend upward in a similar quadratic-looking shape (same Big-O class), but `better_unique`'s curve should sit consistently below `unique`'s - a concrete example of "faster in practice" and "different Big-O complexity" being two separate things.

### Classes and OOP

```python
class Teacher:

    def __init__(self, name, age, experience = 0, tenure = True):
        self.name = name
        self.age = age
        self.experience = experience
        self.tenure = tenure

    def print_info(self):
        string = "{} is a teacher at VIA. He is {} years old and have {} years of experience.".format(self.name, self.age, self.experience)
        if self.tenure:
            string += " They are an associate professor."
        else:
            string += " They are an assistant professor."
        print(string)

    def promote(self):
        if not self.tenure:
            self.tenure = True
        else:
            print("You cannot promote an associate professor!")

    def sayHello(self):
        print("Hello " + self.name)

    def getName(self):
        return self.name
    
```

A standard class definition: `__init__` is the constructor (matches a Java constructor method named after the class), `self` is what `this` is in Java, and `experience = 0` / `tenure = True` are **default parameter values** - Java has no direct equivalent in a single method signature, you'd normally need constructor overloading instead. `print_info` builds its message conditionally based on `self.tenure`.

### Creating Instances

```python
LBS = Teacher("Lars", 55, 22, False)
RSTO = Teacher("Rachel", 35, 0, True)
LBS.print_info()
LBS.promote()
LBS.print_info()
LBS.promote()
LBS.sayHello()
print(LBS.getName())
print(RSTO.getName())
print(LBS.name)
```

Two independent `Teacher` objects, each with their own separate state - calling `.promote()` on `LBS` only changes `LBS`'s `tenure` field, `RSTO` is unaffected. The second `.promote()` call on `LBS` hits the `else` branch and prints a warning instead, since `LBS` is already tenured by then.

### Mutable Default Arguments

```python
def test(something=None):
    if something == None:
        something = set()
        something.add("first")
    something.add(input("Write something: "))
    return something

def myScript():
    first_func_call = set()#test(123)
    second = test(first_func_call)
    print(test(test()))
   # print(test())


myScript()

```

This demonstrates a well-known Python trap: **mutable default arguments**. A default value in a function signature is evaluated only *once*, when the function is defined - not fresh on every call. If you wrote `def test(something=set()):` directly, every call without an argument would share and silently accumulate into the *same* set object across calls. Using `something=None` and creating a fresh `set()` inside the function body when needed sidesteps that entirely. Note this cell calls `input()`, so running it will prompt for keyboard input.
